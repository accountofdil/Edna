## **Stock Price Prediction - Random Forest**

---

```python
import yfinance as yf
import matplotlib.pyplot as plt
from matplotlib import rcParams
import os
import json
import pandas as pd
```

```python
msft = yf.Ticker("MSFT")
msft_hist = msft.history(period = "max")
```

```python
msft_hist
```

```python
msft_hist.head(5)
```

```python
rcParams['font.family'] = 'Times New Roman'

fig, ax = plt.subplots(figsize = (10, 6))

ax.plot(msft_hist.index, msft_hist['Close'], color = 'black', linewidth = 1.5)
ax.set_title("Microsoft ('MSFT') - Close Price", fontweight = 'bold', fontsize = 12)
ax.set_xlabel("Date", fontweight = 'bold')
ax.set_ylabel("Price ($)", fontweight = 'bold')
ax.grid(True, linestyle = '--', alpha = 0.3)

plt.show()
```

```python
data = msft_hist[["Close"]]
data = data.rename(columns = {'Close': 'Actual Close'})
data["Target"] = msft_hist.rolling(2).apply(lambda x: x.iloc[1] > x.iloc[0])["Close"]
```

```python
data.head(5)
```

```python
msft_prev = msft_hist.copy()
msft_prev = msft_prev.shift(1)
msft_prev.head(5)
```

```python
predictors = ["Close", "High", "Low", "Open", "Volume"]
data = data.join(msft_prev[predictors]).iloc[1:]
data.head(5)
```

```python
!pip install scikit-learn
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(n_estimators = 100, min_samples_split = 200, random_state = 1) 
```

```python
train_data = data.iloc[:-100]
test_data = data.iloc[-100:]
```

```python
rf_model.fit(train_data[predictors], train_data["Target"])
```

```python
from sklearn.metrics import precision_score
```

```python
predictions = rf_model.predict(test_data[predictors])
predictions
```

```python
predictions = pd.Series(predictions, index = test_data.index)
predictions
```

```python
precision_score = precision_score(test_data["Target"], predictions)
print(precision_score)
```

```python
combined_data = pd.concat({"Target": test_data["Target"], "Predictions": predictions}, axis = 1)
combined_data
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

combined_data.plot(ax = ax, linewidth = 1.5)
ax.set_title("Microsoft ('MSFT') - Close Price Comparison", fontweight = 'bold', fontsize = 12)
ax.set_xlabel("Date", fontweight = 'bold')
ax.set_ylabel("Price ($)", fontweight = 'bold')
ax.grid(True, linestyle = '--', alpha = 0.3)
ax.legend(title = None)

plt.show()
```

```python
def backtest(data, model, predictors, start = 1000, step = 750):
    
    all_predictions = []
    
    for i in range(start, data.shape[0], step):
        
        train_data = data.iloc[: i].copy()
        test_data = data.iloc[i : i + step].copy()
        
        model.fit(train_data[predictors], train_data["Target"])
        
        predictions = model.predict_proba(test_data[predictors])[:, 1]
        predictions = pd.Series(predictions, index = test_data.index)
        predictions[predictions > 0.6] = 1
        predictions[predictions <= 0.6] = 0
        
        combined_data = pd.concat({"Target": test_data["Target"], "Predictions": predictions}, axis = 1)

        all_predictions.append(combined_data)

    return pd.concat(all_predictions)
```

```python
predictions = backtest(data, rf_model, predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
from sklearn.metrics import precision_score

precision_score_2 = precision_score(predictions["Target"], predictions["Predictions"])
print(precision_score_2)
```

```python
weekly_mean = data.rolling(7).mean()
quarterly_mean = data.rolling(90).mean()
annual_mean = data.rolling(365).mean()
weekly_trend = data.shift(1).rolling(7).mean()["Target"]
```

```python
data["weekly_mean"] = weekly_mean["Close"] / data["Close"]
data["quarterly_mean"] = quarterly_mean["Close"] / data["Close"]
data["annual_mean"] = annual_mean["Close"] / data["Close"]

data["annual_weekly_mean"] = data["annual_mean"] / data["weekly_mean"]
data["annual_quarterly_mean"] = data["annual_mean"] / data["quarterly_mean"]
data["weekly_trend"] = weekly_trend

data["open_close_ratio"] = data["Open"] / data["Close"]
data["high_close_ratio"] = data["High"] / data["Close"]
data["low_close_ratio"] = data["Low"] / data["Close"]
```

```python
full_predictors = predictors + ["weekly_mean", 
                                "quarterly_mean", 
                                "annual_mean", 
                                "annual_weekly_mean", 
                                "annual_quarterly_mean", 
                                "open_close_ratio", 
                                "high_close_ratio", 
                                "low_close_ratio", 
                                "weekly_trend"]
```

```python
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
import numpy as np

tuning_data = data.iloc[365:].dropna()

param_dist = {'criterion': ['gini', 'entropy', 'log_loss'],
              'n_estimators': [50, 100, 200, 300], 
              'min_samples_split': [50, 100, 200, 300],
              'max_depth': [10, 20, 30, None], 
              'min_samples_leaf': [1, 2, 5, 10],
              'max_features': ['sqrt', 'log2', None]}

tscv = TimeSeriesSplit(n_splits = 5)

random_search = RandomizedSearchCV(estimator = RandomForestClassifier(random_state = 1), 
                                   param_distributions = param_dist, 
                                   n_iter = 10, 
                                   scoring = 'precision',
                                   cv = tscv, 
                                   n_jobs = -1, 
                                   random_state = 1)

random_search.fit(tuning_data[full_predictors], tuning_data["Target"])

best_rf_model = random_search.best_estimator_
print("Best parameters found:", random_search.best_params_)
```

```python
predictions = backtest(data.iloc[365:], best_rf_model, full_predictors)
```

```python
from sklearn.metrics import precision_score

precision_score_3 = precision_score(predictions["Target"], predictions["Predictions"])
print(precision_score_3)
```

```python
predictions["Predictions"].value_counts()
```

```python
fig, ax = plt.subplots(figsize = (10, 6))

predictions.iloc[-100:].plot(ax = ax, linewidth = 1.5)
ax.set_title("Microsoft ('MSFT') - Prediction Backtest (Last 100 Observations)", fontweight = 'bold', fontsize = 12)
ax.set_xlabel("Date", fontweight = 'bold')
ax.set_ylabel("Value", fontweight = 'bold')
ax.grid(True, linestyle = '--', alpha = 0.3)
ax.legend(title = None)

plt.show()
```
